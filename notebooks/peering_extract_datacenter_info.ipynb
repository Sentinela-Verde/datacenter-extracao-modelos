```python
import time
import pandas as pd
import requests

BASE_URL = "https://www.peeringdb.com/api"
HEADERS = {"Accept": "application/json"}
REQUEST_DELAY = 1.0
```

```python
params = {"country": "BR", "depth": 2}
resp = requests.get(f"{BASE_URL}/fac", params=params, headers=HEADERS, timeout=30)
resp.raise_for_status()

registros = resp.json()["data"]
print(f"Total de facilities encontradas: {len(registros)}")
```

```python
df = pd.json_normalize(registros, sep="_")

def normalizar_celula(valor):
    if isinstance(valor, list):
        return "; ".join(str(v) for v in valor)
    if isinstance(valor, dict):
        return str(valor)
    return valor

if hasattr(df, "map"):
    df = df.map(normalizar_celula)
else:
    df = df.applymap(normalizar_celula)
```

```python
df.to_csv("facilities_brazil_peering.csv", index=False)
```
