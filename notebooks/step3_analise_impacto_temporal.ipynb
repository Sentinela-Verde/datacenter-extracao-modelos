# Step 3 — Análise temporal e de impacto dos Data Centers

Nesta etapa, o projeto compara os períodos pré, durante e pós implantação do Data Center a partir da série temporal de cobertura gerada no Step 2.

Objetivos:
- calcular variação de vegetação, construção e água;
- comparar pré vs pós;
- gerar indicadores de impacto territorial;
- salvar as métricas para análise e ranking entre sites.

```python
import os
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / 'requirements.txt').exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))
os.chdir(PROJECT_ROOT)

from src.impact import build_impact_report
```

```python
# Parâmetros do Data Center
FACILITY_NAME = 'Ascenty_Vinhedo'
CSV_PATH = 'data/processed/Ascenty_Vinhedo_cobertura_por_ano.csv'
CONSTRUCTION_YEAR = 2020
OPERATION_YEAR = 2024
OUTPUT_DIR = 'data/processed'

print(f'Facility: {FACILITY_NAME}')
print(f'CSV: {CSV_PATH}')
print(f'Construção: {CONSTRUCTION_YEAR} | Operação: {OPERATION_YEAR}')
```

```python
summary, delta = build_impact_report(
    CSV_PATH,
    facility_name=FACILITY_NAME,
    construction_year=CONSTRUCTION_YEAR,
    operation_year=OPERATION_YEAR,
)

print('\n=== RESUMO POR PERÍODO ===')
print(summary.to_string(index=False))

print('\n=== VARIAÇÃO PRÉ X PÓS ===')
print(delta.to_string(index=False))
```

```python
out_dir = Path(OUTPUT_DIR)
out_dir.mkdir(parents=True, exist_ok=True)
summary_path = out_dir / f'{FACILITY_NAME}_impacto_temporal.csv'
delta_path = out_dir / f'{FACILITY_NAME}_delta_pre_pos.csv'

summary.to_csv(summary_path, index=False)
delta.to_csv(delta_path, index=False)

print(f'Arquivo salvo: {summary_path}')
print(f'Arquivo salvo: {delta_path}')
```
